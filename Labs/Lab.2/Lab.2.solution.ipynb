*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
def check_for_win(given_lst):
    if all(cell == 1 for cell in given_lst):
        return 1
    elif all(cell == 2 for cell in given_lst):
        return 2
    else:
        return 0
def outcome(lst):
    rows = lst
    cols = [[row[i] for row in rows] for i in range(len(rows[0]))]
    diag1 = [rows[i][i] for i in range(len(rows))]
    diag2 = [rows[i][len(rows) - 1 - i] for i in range(len(rows))]
    diagonals = [diag1, diag2]
    all_lst = rows + cols + diagonals
    for i in all_lst:
        if check_for_win(i) == 1:
            return 1
        elif check_for_win(i) == 2:
            return 2
    for i in lst:     
        if 0 in i:
            return -1
    return 0
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
def move (board, player, loc):
    row = int(loc[1:])-1
    col = ord(loc[0])-65
    if check_validity(board, player, row, col):
        board[row][col] = player
    else:
        return False, board
    return True, board
````

````python
board = [[2,1,0],[1,1,2],[1,2,1]]
print(move(board, 2, 'C1'))
````

Keep asking for input until the player enters a valid location that results in a valid move.


````python
def user_move(board, player, pl_move=''):
    if pl_move == '':
        loc=input(f"Player {player}'s Move:")
    else:
        loc = pl_move
        print(f"Player {player}'s Move: {loc}")
    row = int(loc[1:])-1
    col = ord(loc[0])-65
    if row > len(board)-1 or col > len(board[0])-1:
        print("It's outside the board. Try again")
        return user_move(board, player)
    else:
        validity, new_board = move(board, player, loc)
        if validity:
            return new_board
        if not validity:
            print("Dude, that place is not empty! Make sure you do a right move this time")
            return user_move(board, player)
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
import copy

def bot1_move(player, board):
    players = [1,2]
    opponent = players[player%2]
    for i_i, i in enumerate(board):
        for j_i, j in enumerate(i):
            if j == 0:
                cp_board = copy.deepcopy(board)
                cp_board[i_i][j_i] = opponent
                if outcome(cp_board) == opponent:
                    return chr(65+j_i)+str(i_i+1)

    for i_i, i in enumerate(board):
        for j_i, j in enumerate(i):
            if j == 0:
                cp_board = copy.deepcopy(board)
                cp_board[i_i][j_i] = player
                if outcome(cp_board) == player:
                    return chr(65+j_i)+str(i_i+1)

    for i_i, i in enumerate(board):
        for j_i, j in enumerate(i):
            if j == 0:
                return chr(65+j_i)+str(i_i+1)

def initiate_game_bot1(n, bot_no):
    players = [1,2]
    board = make_board(n)
    game_loop = True
    turn = 0
    while game_loop:
        game_board(board)
        curr_player = players[turn%2]
        if curr_player != bot_no:
            board = user_move(board, curr_player)
        else:
            board = user_move(board, curr_player, bot1_move(bot_no, board))
            
        turn += 1
        result = outcome(board)
        if result != -1:
            game_loop = False
            game_board(board)
            if result == 1:
                print ("Player 1 wins!")
            elif result == 2:
                print("Player 2 wins!")
            elif result == 0:
                print("It's a tie")
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def scores(board, player, bot_no, depth):
    players = [1,2]
    opponent = players[bot_no%2]
    if outcome(board) == bot_no:
        return 1+depth
    elif outcome(board) == opponent:
        return -1-depth
    elif outcome(board) == 0:
        return 0
    elif depth == 0:
        return 0

    scores_lst = []
    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j]==0:
                cp_board = copy.deepcopy(board)
                cp_board[i][j] = player
                next_player = players[player%2]
                score = scores(cp_board, next_player, bot_no, depth - 1)
                scores_lst.append(score)

    if player == bot_no:
        return max(scores_lst)
    else:
        return min(scores_lst)

def bot2_move(bot_no,board, max_depth = 9):
    players = [1,2]
    best_score = float('-inf')
    best_move = None

    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j]==0:
                cp_board = copy.deepcopy(board)
                cp_board[i][j] = bot_no
                score = scores(cp_board, players[bot_no%2], bot_no, max_depth - 1)
                if score>best_score:
                    best_score = score
                    best_move = chr(65+j)+str(i+1)

    return best_move

def initiate_game_bot2(n, bot_no):
    players = [1,2]
    board = make_board(n)
    game_loop = True
    turn = 0
    while game_loop:
        game_board(board)
        curr_player = players[turn%2]
        if curr_player != bot_no:
            board = user_move(board, curr_player)
        else:
            board = user_move(board, curr_player, bot2_move(bot_no, board))
            
        turn += 1
        result = outcome(board)
        if result != -1:
            game_loop = False
            game_board(board)
            if result == 1:
                print ("Player 1 wins!")
            elif result == 2:
                print("Player 2 wins!")
            elif result == 0:
                print("It's a tie")
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
def bot_move(board, player, loc):
    _, new_board = move(board, player, loc)
    return new_board
    
def bot_vs_bot_simulator(n, first_bot_no, starting_move, max_depth):
    bots = [1,2]
    board = make_board(n)
    curr_player = first_bot_no
    board = bot_move(board, curr_player, starting_move)
    game_loop = True
    while game_loop:
        curr_player = bots[curr_player%2]
        if curr_player == 2:
            board = bot_move(board, curr_player, bot2_move(curr_player, board, max_depth))
        else:
            board = bot_move(board, curr_player, bot1_move(curr_player, board))
            
        result = outcome(board)
        if result != -1:
            game_loop = False
            return result

def relative_openings(n): #made so that the bots use different opening moves all the time so that the games are not the same
    lst = []
    lst.append(chr(65+n//2)+str(n//2+1)) #centerish
    lst.append(chr(65)+str(1)) #top-left corner
    lst.append(chr(64+n)+str(n)) #bottom-right corner
    lst.append(chr(65)+str(n//2+1)) #left edgish
    lst.append(chr(65+n//2)+str(1))#top edgish
    return lst

def bots_tournament(n, max_depth):
    openings = relative_openings(n)
    matches = []
    for i in openings:
        matches.append(bot_vs_bot_simulator(n, 1, i, max_depth))
        matches.append(bot_vs_bot_simulator(n, 2, i, max_depth))
    bot_2_win_rate = str((matches.count(2)/10)*100)+'%'
    tie_rate = str((matches.count(0)/10)*100)+'%'
    return matches, bot_2_win_rate, tie_rate
````

````python
# the win rate for the "smarter" model's
matches, win_rate, tie_rate = bots_tournament(3, 9)
print(matches)
print(win_rate)
print(tie_rate)
````
